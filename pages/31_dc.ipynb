# 2. זרם ישר, ניו יורק 1882

**זמן לעבודה: 10 דקות**

ב-1882 נפתחה בניו יורק אחת מתחנות הכוח הציבוריות הראשונות. היא סיפקה **זרם ישר** במתח 110 וולט למנורות בבתים שסביבה, דרך חוטי נחושת.

לחוט נחושת יש התנגדות, ולכן חלק מהמתח "הולך לאיבוד" בדרך. ככל שהבית רחוק יותר מהתחנה, המתח שמגיע למנורות נמוך יותר. ההתנגדות של חוט באורך <span dir="ltr">&#8288;$L$</span> ובשטח חתך <span dir="ltr">&#8288;$A$</span> היא <span dir="ltr">&#8288;$R = \rho L / A$</span>.   <span dir="ltr">&#8288;$\rho$</span> הוא ההתנגדות הסגולית של החומר. הזרם עובר בחוט אל הבית ובחוט שני בחזרה, ולכן לבית במרחק <span dir="ltr">&#8288;$d$</span> אורך החוט הוא <span dir="ltr">&#8288;$L = 2d$</span>. לפי חוק אוהם, המתח שמגיע למנורות הוא:

<div dir="ltr">

$$V = V_{station} - I \cdot R = V_{station} - I \cdot \frac{\rho \cdot 2d}{A}$$

</div>

בתא הנתונים מופיעים נתוני הרשת ופונקציה שמחפשת את המרחק הגדול ביותר שבו המנורות עדיין דולקות כראוי. שימו לב: שטח החתך נתון בממ"ר, ו-1 ממ"ר הוא <span dir="ltr">&#8288;$10^{-6}$</span> מ"ר.

</div>

````python
V_station = 110     # voltage at the station (V)
V_min = 100         # the lamps need at least this voltage (V)
I = 50              # current drawn by one street (A)
rho = 1.7e-8        # resistivity of copper (ohm*m)
area_mm2 = 100      # cross-section of each copper wire (mm^2)
step = 50           # the distances are checked every 50 m


# the function as it was given to you: the farthest distance (m), in steps of `step`, where the lamps get at least V_min
# (it calls voltage_at, which you will write in section 1)
def max_distance():
    d = 0
    while voltage_at(d) >= V_min:
        d = d + step
    return d
````
