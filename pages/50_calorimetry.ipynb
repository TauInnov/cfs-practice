ב-1819 מדדו הפיזיקאים פייר לואי דולון ואלכסי תרז פטי את **החום הסגולי** של מתכות רבות: כמות האנרגיה שדרושה כדי לחמם גרם אחד של חומר במעלה אחת. לכל מתכת יש חום סגולי משלה, ולכן אפשר לזהות בעזרתו ממה עשוי גוף.

במעבדת ההוראה יש חמש קוביות מתכת, A עד E, והתוויות עם שם המתכת נפלו מהן. כל קובייה חוממה במים רותחים ל-<span dir="ltr">&#8288;$T_{{hot}} = 100$</span>°C והוכנסה לכוס מבודדת עם <span dir="ltr">&#8288;$m_w = 100$</span> גרם מים בטמפרטורה <span dir="ltr">&#8288;$T_w = 20$</span>°C. כשהטמפרטורה התייצבה, נמדדה הטמפרטורה הסופית <span dir="ltr">&#8288;$T_f$</span>. החום שמסרה הקובייה שווה לחום שקיבלו המים (את החום שקיבלה הכוס מזניחים), ומכאן החום הסגולי של קובייה במסה <span dir="ltr">&#8288;$m$</span>:

<div dir="ltr">

$$c = \frac{m_w \cdot c_w \cdot (T_f - T_w)}{m \cdot (T_{hot} - T_f)}$$

</div>

כאשר <span dir="ltr">&#8288;$c_w$</span> הוא החום הסגולי של מים. בתא הנתונים מופיעים הקבועים, מילון של החום הסגולי של חמש מתכות, קוד שיוצר את קובץ המדידות `samples.txt`, ופונקציה שקוראת אותו.

</div>

```python
c_w = 4.186       # specific heat of water (J/(g*K))
m_w = 100.0       # mass of the water in the cup (g)
T_w = 20.0        # initial temperature of the water (C)
T_hot = 100.0     # initial temperature of every cube, heated in boiling water (C)

# specific heat (J/(g*K)) of five metals, by chemical symbol
c_table = {"Al": 0.897, "Fe": 0.449, "Cu": 0.385, "Ag": 0.235, "Pb": 0.129}

# the lab notebook: one line per cube, its mass (g) and the final temperature of the water (C)
with open("samples.txt", "w") as f:
    f.write("sample,mass_g,T_final_C\n")
    f.write("A,150.0,29.7\n")
    f.write("B,250.0,25.7\n")
    f.write("C,60.0,29.1\n")
    f.write("D,120.0,29.1\n")
    f.write("E,200.0,28.1\n")


# the function as it was given to you: a dict from each cube's letter to a tuple (mass, T_final)
def read_samples(path):
    samples = {}
    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            samples[parts[0]] = (parts[1], parts[2])
    return samples
```

```python
# the function as it was given to you: the metal in c_table whose specific heat is closest to c
def closest_metal(c):
    best_symbol = ""
    best_diff = 1000
    for symbol, c_metal in c_table.items():
        diff = abs(c - c_metal)
        if diff < best_diff:
            best_symbol = symbol
            best_diff = diff
    return best_symbol, best_diff

# TODO: answer (b) before running, then build the string metals
```
