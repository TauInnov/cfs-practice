ב-1827 פרסם גאורג אוהם, מורה לפיזיקה בקלן, שהזרם בתיל פרופורציונלי למתח שמפעילים עליו: <span dir="ltr">&#8288;$I = V / R$</span>. היום ההתנגדות <span dir="ltr">&#8288;$R$</span> היא רכיב קטן, **נגד**, והערך שלו מסומן עליו בפסי צבע. לכל צבע יש ספרה: שני הפסים הראשונים הם שתי הספרות של הערך, והפס השלישי הוא חזקה של 10:

<div dir="ltr">

$$R = (10 \cdot d_1 + d_2) \cdot 10^{d_3} \quad \text{(אוהם)}$$

</div>

נגד שמחובר למתח <span dir="ltr">&#8288;$V$</span> מתחמם בהספק <span dir="ltr">&#8288;$P = V^2 / R$</span>, ונגד קטן רגיל עומד בהספק של עד 0.25 וואט בלי להתחמם יותר מדי. כשמחברים כמה נגדים **במקביל** לסוללה, כל אחד מהם מקבל את מלוא המתח, וההתנגדות השקולה שלהם היא:

<div dir="ltr">

$$\frac{1}{R_{eq}} = \frac{1}{R_1} + \frac{1}{R_2} + \dots, \qquad I = \frac{V}{R_{eq}}$$

</div>

בתא הנתונים מופיעים מתח הסוללה, מילון של הספרה של כל צבע, קוד שיוצר את הקובץ `kit.txt` עם הצבעים של חמשת הנגדים בערכה, ושתי פונקציות שקיבלתם.

</div>

```python
V = 9.0           # battery voltage (V)
P_max = 0.25      # the most power a small resistor can take without overheating (W)

# the digit of each color in the color code
color_digit = {"black": 0, "brown": 1, "red": 2, "orange": 3, "yellow": 4,
               "green": 5, "blue": 6, "violet": 7, "gray": 8, "white": 9}

# the kit: one line per resistor, its letter and the colors of its three bands
with open("kit.txt", "w") as f:
    f.write("A, brown, black, red\n")
    f.write("B, red, red, brown\n")
    f.write("C, yellow, violet, brown\n")
    f.write("D, gray, red, brown\n")
    f.write("E, orange, orange, red\n")


# the functions as they were given to you
# a dict from each resistor's letter to a tuple of the colors of its three bands
def read_kit(path):
    kit = {}
    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            kit[parts[0]] = (parts[1], parts[2], parts[3])
    return kit


# the resistance (ohm) of a resistor, from the tuple of its three colors
def resistance(colors):
    d1 = color_digit[colors[0]]
    d2 = color_digit[colors[1]]
    d3 = color_digit[colors[2]]
    return (10 * d1 + d2) * 10 * d3
```
