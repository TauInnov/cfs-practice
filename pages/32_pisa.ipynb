לפי סיפור מפורסם, גלילאו הפיל שני כדורים במשקלים שונים מראש מגדל פיזה, כדי להראות שהם פוגעים בקרקע יחד. בלי אוויר זה נכון: כל הגופים נופלים בתאוצה <span dir="ltr">&#8288;$g$</span>. עם אוויר, כוח החיכוך מאט כדור קל יותר מכדור כבד כאשר שניהם באותו גודל.

בתרגיל נחשב את הנפילה **צעד אחר צעד**: מחלקים את הזמן לצעדים קטנים באורך `dt`, ובכל צעד מעדכנים את התאוצה, המהירות והגובה:

<div dir="ltr">

$$a = g - c \cdot v^2, \qquad v \leftarrow v + a \cdot dt, \qquad y \leftarrow y - v \cdot dt, \qquad t \leftarrow t + dt$$

</div>

כאן <span dir="ltr">&#8288;$v$</span> היא המהירות כלפי מטה, ו-<span dir="ltr">&#8288;$c$</span> הוא מקדם החיכוך של הכדור (חיכוך האוויר חלקי המסה). בתא הנתונים מופיעים ארבעה כדורים באותו גודל (קוטר 10 ס"מ) מחומרים שונים, ופונקציה שמחשבת את זמן הנפילה.

</div>

````python
g = 9.81          # gravitational acceleration (m/s^2)
h = 56            # height of the tower (m)
dt = 0.001        # time step (s)

# four balls of the same size (10 cm), and their drag coefficient c (1/m): the drag acceleration is c * v**2
materials = ["iron", "marble", "wood", "cork"]
c_values = [0.00054, 0.00157, 0.00705, 0.01762]
c_iron = c_values[0]
c_wood = c_values[2]


# the function as it was given to you: the time (s) for a ball with drag coefficient c to fall from height h
def fall_time(c):
    y = h
    v = 0
    t = 0
    while y < 0:
        a = g - c * v**2
        v = v + a * dt
        y = y - v * dt
        t = t + dt
    return t
````
