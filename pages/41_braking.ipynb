נהג נוסע במהירות <span dir="ltr">&#8288;$v_0$</span> ורואה מכשול במרחק <span dir="ltr">&#8288;$D$</span> לפניו. מהרגע שהוא רואה את המכשול ועד שהרכב עוצר, יש שני שלבים:

<p align="center"></p>

1. **זמן התגובה**, <span dir="ltr">&#8288;$t_r$</span>: הזמן שעובר עד שהנהג לוחץ על הבלמים. בזמן הזה הרכב ממשיך במהירות קבועה <span dir="ltr">&#8288;$v_0$</span>, ולכן עובר מרחק <span dir="ltr">&#8288;$v_0 \, t_r$</span>.
2. **הבלימה**: כוח החיכוך בין הצמיגים לכביש, <span dir="ltr">&#8288;$\mu m g$</span>, מאט את הרכב. לפי החוק השני של ניוטון גודל התאוצה הוא <span dir="ltr">&#8288;$\mu g$</span>, בכיוון הפוך לתנועה: היא קבועה, ולא תלויה במסת הרכב. מהקשר <span dir="ltr">&#8288;$v^2 = v_0^2 - 2 \mu g \, x$</span>, עם מהירות סופית <span dir="ltr">&#8288;$v = 0$</span>, מקבלים את מרחק הבלימה <span dir="ltr">&#8288;$v_0^2 / (2 \mu g)$</span>.

<p align="center"></p>

מרחק העצירה הכולל הוא סכום שני המרחקים:

<div dir="ltr">

$$d = v_0 \, t_r + \frac{v_0^2}{2 \mu g}$$

</div>

**הגדלים בתרגיל**

- <span dir="ltr">&#8288;$v_0$</span>: המהירות של הרכב ברגע שהנהג רואה את המכשול, במטר לשנייה. ברשימה `speeds_kmh` המהירויות בקמ"ש: כדי להמיר קמ"ש למטר לשנייה מחלקים ב-3.6, ובכיוון ההפוך מכפילים ב-3.6.
- <span dir="ltr">&#8288;$t_r$</span>: זמן התגובה של הנהג, שנייה אחת.
- <span dir="ltr">&#8288;$\mu$</span>: **מקדם החיכוך** בין הצמיגים לכביש, מספר בלי יחידות. ככל שהוא גדול יותר, החיכוך חזק יותר, והרכב עוצר במרחק קצר יותר. בכביש יבש <span dir="ltr">&#8288;$\mu = 0.7$</span>, ובכביש רטוב <span dir="ltr">&#8288;$\mu = 0.4$</span>.
- <span dir="ltr">&#8288;$g$</span>: תאוצת הכובד, 9.81 m/s².
- <span dir="ltr">&#8288;$D$</span>: המרחק למכשול ברגע שהנהג רואה אותו, 40 מטר. הרכב עוצר בזמן אם <span dir="ltr">&#8288;$d \le D$</span>.

**הנחות.** הכביש מישורי, זמן התגובה קבוע, והתאוצה בזמן הבלימה קבועה.

</div>

```python
g = 9.81                        # gravitational acceleration (m/s^2)
t_react = 1.0                   # reaction time of the driver (s)
mu_dry = 0.7                    # friction coefficient between tires and road, dry road
mu_wet = 0.4                    # friction coefficient between tires and road, wet road
speeds_kmh = [30, 50, 70, 90]   # speeds to check (km/h)
D = 40.0                        # distance to the obstacle when the driver sees it (m)

mu = mu_dry                     # the road is dry


# the function as it was given to you: the stopping distance (m) for an initial speed v0 (m/s)
def stopping_distance(v0, mu=mu):
    return v0 * t_react + v0 ** 2 / (2 * mu * g)


# a new list with f applied to every value
def apply_to_all(f, values):
    return [f(x) for x in values]
```

**4. מנוסחה לפונקציה: המהירות המקסימלית.**

ככל שהמהירות גדולה יותר, מרחק העצירה גדול יותר. לכן המהירות המקסימלית היא המהירות שבה מרחק העצירה שווה **בדיוק** ל-<span dir="ltr">&#8288;$D$</span>. בכל מהירות גבוהה ממנה, הרכב כבר לא עוצר בזמן. מציבים <span dir="ltr">&#8288;$d = D$</span> בנוסחה, ומעבירים את <span dir="ltr">&#8288;$D$</span> לאגף שמאל:

<div dir="ltr">

$$\frac{1}{2 \mu g} \, v_0^2 + t_r \, v_0 - D = 0$$

</div>

זו משוואה ריבועית במשתנה <span dir="ltr">&#8288;$v_0$</span>, מהצורה <span dir="ltr">&#8288;$a v_0^2 + b v_0 + c = 0$</span>, עם המקדמים:

<div dir="ltr">

$$a = \frac{1}{2 \mu g}, \qquad b = t_r, \qquad c = -D$$

</div>

לפי נוסחת השורשים, הפתרונות שלה הם:

<div dir="ltr">

$$v_0 = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

</div>

מכיוון ש-<span dir="ltr">&#8288;$c$</span> שלילי, אחד הפתרונות חיובי והשני שלילי. רק לפתרון החיובי יש משמעות פיזיקלית: המהירות של רכב שנוסע קדימה.
