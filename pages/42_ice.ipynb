# 3. קרח בכוס

**זמן לעבודה: 10 דקות**

בכוס יש מים. מכניסים לכוס קוביות קרח, ומחכים עד שהכוס מגיעה ל**שיווי משקל תרמי**: הטמפרטורה בכל הכוס אחידה, ואינה משתנה יותר. המים מתקררים, והחום שהם מוסרים מתיך את הקרח. השאלה היא אם כל הקרח מותך, או שחלק ממנו נשאר.

<p align="center"></p>

**הגדלים בתרגיל**

- <span dir="ltr">&#8288;$m_w$</span>: מסת המים בכוס, בגרמים.
- <span dir="ltr">&#8288;$T_w$</span>: הטמפרטורה ההתחלתית של המים, במעלות צלזיוס.
- <span dir="ltr">&#8288;$m_i$</span>: מסת הקרח, בגרמים. הקרח נכנס לכוס בטמפרטורה של 0°C בדיוק.
- <span dir="ltr">&#8288;$c$</span>: **החום הסגולי של מים**, 4.186 J/(g·°C). זו כמות האנרגיה, בג'אול, שצריך כדי לחמם גרם אחד של מים במעלה אחת. גרם מים שמתקרר במעלה אחת מוסר את אותה כמות אנרגיה. לכן מסה <span dir="ltr">&#8288;$m$</span> של מים שהטמפרטורה שלה משתנה ב-<span dir="ltr">&#8288;$\Delta T$</span> מעלות מוסרת או קולטת <span dir="ltr">&#8288;$m \, c \, \Delta T$</span> ג'אול.
- <span dir="ltr">&#8288;$L$</span>: **חום ההתכה של קרח**, 334 J/g. זו כמות האנרגיה שצריך כדי להתיך גרם אחד של קרח ב-0°C ולהפוך אותו למים ב-0°C. בזמן ההתכה הטמפרטורה לא משתנה. לכן ההתכה של <span dir="ltr">&#8288;$m_i$</span> גרם קרח דורשת <span dir="ltr">&#8288;$m_i L$</span> ג'אול.

**הנחות.** הכוס והאוויר לא קולטים ולא מוסרים חום: כל החום שהמים מוסרים עובר לקרח. בנוסף, בשלב הזה **מניחים שכל הקרח מותך**.

**הנוסחה לטמפרטורה הסופית.** החום שהמים מוסרים כשהם מתקררים מ-<span dir="ltr">&#8288;$T_w$</span> ל-<span dir="ltr">&#8288;$T_f$</span> שווה לחום שהקרח קולט. הקרח קולט חום פעמיים: כדי להיות מותך, ואחר כך כדי שהמים שנוצרו ממנו יתחממו מ-0°C ל-<span dir="ltr">&#8288;$T_f$</span>:

<div dir="ltr">

$$m_w \, c \, (T_w - T_f) = m_i L + m_i \, c \, (T_f - 0)$$

</div>

מבודדים את <span dir="ltr">&#8288;$T_f$</span> ומקבלים:

<div dir="ltr">

$$T_f = \frac{m_w \, c \, T_w - m_i L}{(m_w + m_i) \, c}$$

</div>

**התנאי: מתי מותר להשתמש בנוסחה?**

הנוסחה נכונה **רק אם כל הקרח אכן מותך**. כדי לדעת אם זה קורה, משווים שתי כמויות חום:

- <span dir="ltr">&#8288;$m_w \, c \, T_w$</span>: החום **המקסימלי** שהמים יכולים למסור. זה החום שהם מוסרים כשהם מתקררים עד 0°C. כל עוד יש בכוס קרח, המים לא מתקררים מתחת ל-0°C.
- <span dir="ltr">&#8288;$m_i L$</span>: החום **הדרוש** כדי להתיך את כל הקרח.

<div dir="ltr">

$$m_w \, c \, T_w \;\geq\; m_i L$$

</div>

- **אם אי השוויון מתקיים**, יש מספיק חום: כל הקרח מותך, והחום שנשאר מחמם את המים בכוס מעל 0°C. הנוסחה ל-<span dir="ltr">&#8288;$T_f$</span> נכונה.
- **אם אי השוויון לא מתקיים**, אין מספיק חום: המים מתקררים עד 0°C בדיוק, רק חלק מהקרח מותך, והטמפרטורה הסופית היא <span dir="ltr">&#8288;$T_f = 0$</span>. במקרה הזה **הנוסחה ל-<span dir="ltr">&#8288;$T_f$</span> לא נכונה**, ונותנת מספר חסר משמעות.

**לכן, לפני כל שימוש בנוסחה ל-<span dir="ltr">&#8288;$T_f$</span>, צריך לבדוק את אי השוויון.**

<p align="center"></p>

</div>

```python
c = 4.186                               # specific heat of water: energy to heat 1 g of water by 1 C (J/(g*C))
L = 334.0                               # heat of fusion of ice: energy to melt 1 g of ice at 0 C (J/g)
m_water = 250.0                         # mass of the water in each glass (g)
m_cube = 20.0                           # mass of one ice cube, at 0 C (g)
T_glasses = [20.0, 30.0, 40.0, 50.0]    # initial temperature of the water in four glasses (C)


# a new list with f applied to every value
def apply_to_all(f, values):
    return [f(x) for x in values]
```
