שמים קוביות קרח בטמפרטורה של 0°C בכוס מים. המים מתקררים, והחום שהם מוסרים ממיס את הקרח: כדי להמיס גרם אחד של קרח צריך <span dir="ltr">&#8288;$L$</span> ג'אול (חום ההיתוך). אם כל הקרח נמס, הטמפרטורה הסופית של המים בכוס היא:

<div dir="ltr">

$$T_f = \frac{m_w c \, T_w - m_i L}{(m_w + m_i) \, c}$$

</div>

כאן <span dir="ltr">&#8288;$m_w$</span> ו-<span dir="ltr">&#8288;$T_w$</span> הם המסה והטמפרטורה ההתחלתית של המים, <span dir="ltr">&#8288;$m_i$</span> היא מסת הקרח, ו-<span dir="ltr">&#8288;$c$</span> הוא החום הסגולי של המים.

הנוסחה נכונה רק אם למים יש מספיק חום כדי להמיס את כל הקרח. כשהמים מתקררים עד 0°C הם מוסרים <span dir="ltr">&#8288;$m_w c \, T_w$</span> ג'אול, ולכן התנאי הוא <span dir="ltr">&#8288;$m_w c \, T_w \ge m_i L$</span>. אחרת, חלק מהקרח נשאר, והטמפרטורה הסופית היא 0°C.

</div>

```python
c = 4.186                               # specific heat of water (J/(g*C))
L = 334.0                               # heat of fusion of ice (J/g)
m_water = 250.0                         # mass of the water in each glass (g)
m_cube = 20.0                           # mass of one ice cube, at 0 C (g)
T_glasses = [20.0, 30.0, 40.0, 50.0]    # initial temperature of the water in four glasses (C)


# a new list with f applied to every value
def apply_to_all(f, values):
    return [f(x) for x in values]
```

```python
# the code as it was given to you: one ice cube in each glass, and how much each glass cooled
T_final = T_glasses
for i in range(len(T_final)):
    T_final[i] = (m_water * c * T_final[i] - m_cube * L) / ((m_water + m_cube) * c)

cooling = [T_glasses[i] - T_final[i] for i in range(len(T_glasses))]
print(T_final)
print(cooling)
```
