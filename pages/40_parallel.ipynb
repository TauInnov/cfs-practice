ארבעה נגדים מחוברים במקביל לסוללה של 12 וולט. לכל נגד אותו מתח <span dir="ltr">&#8288;$V$</span>, ולכן לפי חוק אוהם:

<div dir="ltr">

$$I_i = \frac{V}{R_i}, \qquad \frac{1}{R_p} = \sum_i \frac{1}{R_i}, \qquad P_i = \frac{V^2}{R_i}$$

</div>

<span dir="ltr">&#8288;$I_i$</span> הוא הזרם בנגד <span dir="ltr">&#8288;$i$</span>, <span dir="ltr">&#8288;$R_p$</span> היא ההתנגדות השקולה, ו-<span dir="ltr">&#8288;$P_i$</span> הוא ההספק שמתפזר בנגד. בסעיפים 1 עד 3 הסוללה אידיאלית, ובסעיף 4 נוסיף לה התנגדות פנימית.

</div>

```python
V = 12.0                          # battery voltage (V)
resistors = [100, 220, 470, 1000]   # the resistors, all connected in parallel to the battery (ohm)
r_internal = 2.0                    # internal resistance of the battery (ohm)


# the function as it was given to you: the equivalent resistance of resistors in parallel
def parallel(rs):
    for i in range(len(rs)):
        rs[i] = 1 / rs[i]
    return 1 / sum(rs)


# the equivalent resistance of resistors in series
def series(rs):
    return sum(rs)


# a new list with f applied to every value
def apply_to_all(f, values):
    return [f(x) for x in values]
```
