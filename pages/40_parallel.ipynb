ארבעה נגדים מחוברים **במקביל** לסוללה: כל נגד מחובר ישירות לשני ההדקים של הסוללה, כמו בשרטוט.

<p align="center"></p>

**הגדלים בתרגיל**

- <span dir="ltr">&#8288;$V$</span>: המתח על הנגדים, בוולט (V). בחיבור במקביל יש על כל הנגדים **אותו מתח**, המתח של הסוללה: 12 וולט.
- <span dir="ltr">&#8288;$R_i$</span>: ההתנגדות של נגד מספר <span dir="ltr">&#8288;$i$</span>, באוהם (Ω). הערכים נמצאים ברשימה `resistors`.
- <span dir="ltr">&#8288;$I_i$</span>: הזרם דרך נגד <span dir="ltr">&#8288;$i$</span>, באמפר (A). לפי חוק אוהם <span dir="ltr">&#8288;$I_i = V/R_i$</span>: ככל שההתנגדות קטנה יותר, הזרם גדול יותר.
- <span dir="ltr">&#8288;$I$</span>: הזרם הכולל שיוצא מהסוללה. לפי חוק הצמתים של קירכהוף הוא מתפצל בין הנגדים, ולכן <span dir="ltr">&#8288;$I = I_1 + I_2 + I_3 + I_4$</span>.
- <span dir="ltr">&#8288;$P_i$</span>: ההספק בנגד <span dir="ltr">&#8288;$i$</span>, בוואט (W): כמה אנרגיה חשמלית הופכת בנגד לחום בכל שנייה. <span dir="ltr">&#8288;$P_i = V \cdot I_i = V^2 / R_i$</span>.

**ההתנגדות השקולה.** ארבעת הנגדים יחד מתנהגים כמו נגד אחד, <span dir="ltr">&#8288;$R_p$</span>, שמושך מהסוללה את אותו זרם כולל: <span dir="ltr">&#8288;$I = V/R_p$</span>. מציבים <span dir="ltr">&#8288;$I = \sum_i V/R_i$</span>, ומקבלים:

<div dir="ltr">

$$\frac{1}{R_p} = \sum_i \frac{1}{R_i}$$

</div>

ההתנגדות השקולה במקביל תמיד **קטנה** מההתנגדות הקטנה ביותר, כי כל נגד נוסף פותח לזרם מסלול נוסף.

להשוואה, אם היו מחברים את אותם נגדים **בטור**, זה אחרי זה, ההתנגדות השקולה הייתה הסכום שלהם: <span dir="ltr">&#8288;$R_s = \sum_i R_i$</span>.

**הנחות.** לחוטים אין התנגדות. בסעיפים 1 עד 3 הסוללה אידיאלית (<span dir="ltr">&#8288;$r = 0$</span> בשרטוט), ולכן המתח על הנגדים הוא תמיד 12 וולט. בסעיף 4 נוסיף לסוללה התנגדות פנימית <span dir="ltr">&#8288;$r$</span>.

</div>

```python
V = 12.0                            # battery voltage (V)
resistors = [100, 220, 470, 1000]   # the resistances, all connected in parallel to the battery (ohm)
r_internal = 2.0                    # internal resistance of the battery, used in section 4 (ohm)


# the function as it was given to you: the equivalent resistance of resistors in parallel
def parallel(rs):
    for i in range(len(rs)):
        rs[i] = 1 / rs[i]
    return 1 / sum(rs)


# the equivalent resistance of resistors in series
def series(rs):
    return sum(rs)


# a new list with f applied to every value
def apply_to_all(f, values):
    return [f(x) for x in values]
```

**4. סוללה אמיתית: מנוסחה לפונקציה.**

עד עכשיו הנחנו שהסוללה אידיאלית. לסוללה אמיתית יש **התנגדות פנימית** <span dir="ltr">&#8288;$r$</span>: ההתנגדות של החומרים שבתוך הסוללה עצמה. בשרטוט היא מופיעה בתוך הקופסה המקווקוות.

- <span dir="ltr">&#8288;$\varepsilon$</span>: **הכא"מ** (הכוח האלקטרו-מניע) של הסוללה: המתח שלה כשלא עובר בה זרם. כאן 12 וולט.
- <span dir="ltr">&#8288;$r$</span>: ההתנגדות הפנימית. בשאלה, 2 אוהם (`r_internal`).

הזרם עובר דרך <span dir="ltr">&#8288;$r$</span> ואחר כך דרך הנגדים שבמקביל, ולכן <span dir="ltr">&#8288;$r$</span> מחוברת **בטור** לנגד השקול <span dir="ltr">&#8288;$R_p$</span>, וההתנגדות הכוללת במעגל היא <span dir="ltr">&#8288;$r + R_p$</span>. החישוב נעשה בשלושה שלבים:

**שלב 1: הזרם מהסוללה**, לפי חוק אוהם על כל המעגל:

<div dir="ltr">

$$I = \frac{\varepsilon}{r + R_p}$$

</div>

**שלב 2: המתח שמגיע לנגדים** (מתח ההדקים). חלק מהמתח, <span dir="ltr">&#8288;$I \cdot r$</span>, "הולך לאיבוד" על ההתנגדות הפנימית, ורק השאר מגיע לנגדים:

<div dir="ltr">

$$V_t = \varepsilon - I \cdot r$$

</div>

**שלב 3: ההספק.** על כל נגד יש עכשיו מתח <span dir="ltr">&#8288;$V_t$</span> במקום <span dir="ltr">&#8288;$\varepsilon$</span>, ולכן:

<div dir="ltr">

$$P_i = \frac{V_t^2}{R_i}, \qquad P_{total} = \sum_i P_i$$

</div>

כשהסוללה אידיאלית (<span dir="ltr">&#8288;$r = 0$</span>) מתקבל <span dir="ltr">&#8288;$V_t = \varepsilon$</span>, ואלה בדיוק הנוסחאות של סעיפים 1 עד 3.

בכל הפונקציות בסעיף הזה יש אותם שלושה קלטים: `rs`, רשימת ההתנגדויות של הנגדים שבמקביל; `emf`, הכא"מ <span dir="ltr">&#8288;$\varepsilon$</span>, עם ערך ברירת מחדל 12; ו-`r`, ההתנגדות הפנימית, עם ערך ברירת מחדל 0, כלומר סוללה אידיאלית.

- (א) כתבו פונקציה `battery_current(rs, emf=12.0, r=0.0)` שמחזירה את <span dir="ltr">&#8288;$I$</span> לפי שלב 1. את <span dir="ltr">&#8288;$R_p$</span> חשבו בעזרת `parallel` שתיקנתם בסעיף 2.
- (ב) כתבו פונקציה `terminal_voltage(rs, emf=12.0, r=0.0)` שמחזירה את <span dir="ltr">&#8288;$V_t$</span> לפי שלב 2. את <span dir="ltr">&#8288;$I$</span> היא מחשבת בעזרת `battery_current`, עם אותם `emf` ו-`r` שהיא עצמה קיבלה.
- (ג) בדקו בשני ביטויים בוליאניים: בסוללה אידיאלית, `terminal_voltage(resistors)` שווה בדיוק ל-12, כי אין מתח שהולך לאיבוד; ועם ההתנגדות הפנימית, `terminal_voltage(resistors, r=r_internal)` קטן מ-12.
- (ד) כתבו פונקציה `total_power(rs, emf=12.0, r=0.0)` שמחזירה את <span dir="ltr">&#8288;$P_{total}$</span> לפי שלב 3: היא מחשבת את <span dir="ltr">&#8288;$V_t$</span> בעזרת `terminal_voltage`, ואחר כך את סכום ההספקים בכל הנגדים ב-`rs`.
- (ה) **בדיקה לסוללה אידיאלית.** בלי התנגדות פנימית, ההספק הכולל הוא <span dir="ltr">&#8288;$V^2/R_p$</span>, עם `R_p` מסעיף 2. בדקו בביטוי בוליאני ש-`total_power(resistors)` שווה לו, עד הפרש של `1e-12`.
