**שלב אחר שלב**

הסעיפים כאן זהים לסעיפי העמוד. סעיפים 1 ו-2 בודקים אילו גופים קשורים, סעיף 3 מזהה את הגוף הבין-כוכבי, וסעיפים 4 ו-5 מחשבים את המהירות.

**1. מהירות המילוט.** `GM_sun` נתון בקילומטרים, ולכן ממירים את המרחק מיחידות אסטרונומיות לקילומטרים:

```python
def escape_speed(r_au):
    return (2 * GM_sun / (r_au * AU)) ** 0.5

escape_speed(1.0)    # about 42.1 km/s
```

**2. קשור או לא.** הביטוי הבוליאני עצמו הוא התשובה:

```python
def is_bound(r_au, v):
    return v < escape_speed(r_au)

b1 = is_bound(r1, v1)    # True:  29.78 < 42.12
b2 = is_bound(r2, v2)    # True:  54.55 < 55.02, barely
b3 = is_bound(r3, v3)    # False: 17.0 > 3.83
b4 = is_bound(r4, v4)    # False: 87.3 > 83.25
```

**3. הגוף הבין-כוכבי.** שני גופים אינם קשורים, אבל אחד מהם מעשה ידי אדם:

```python
def is_interstellar(r_au, v, artificial):
    return not is_bound(r_au, v) and not artificial

is_interstellar(r1, v1, art1)    # False
is_interstellar(r2, v2, art2)    # False
is_interstellar(r3, v3, art3)    # False: unbound, but made by humans
is_interstellar(r4, v4, art4)    # True: 'Oumuamua
```

**4. המהירות הרחק מהשמש.** הפונקציה קוראת לשתי הפונקציות הקודמות, ובכל מסלול יש `return`:

```python
def v_infinity(r_au, v):
    if is_bound(r_au, v):
        return 0.0
    else:
        return (v**2 - escape_speed(r_au)**2) ** 0.5
```

**5. התשובה.** זו התשובה שמגישים:

```python
v_inf = round(v_infinity(r4, v4), 1)    # 26.3 (km/s)
```
