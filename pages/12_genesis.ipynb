```python
def deploy_intended(a):
    return -a >= a_threshold    # the same as a <= -a_threshold
```

**2. המתג שהותקן הפוך.** אותה פונקציה, על הערך בסימן ההפוך:

```python
def deploy_as_built(a):

```

d1 = deploy_intended(a1) != deploy_as_built(a1)    # False: both False
