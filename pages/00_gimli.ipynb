**התשובה: כ-10,117 ק"ג דלק בהמראה**, פחות ממחצית מ-22,300 הק"ג שנדרשו.

**הנתונים**

הפרמטרים מתא הנתונים בעמוד, שעליהם מבוססות כל השורות בהמשך:

```python
fuel_in_tanks_L = 7682     # fuel measured in the tanks before refuelling (liters)
required_kg = 22300        # fuel required for the flight (kg)
factor_used = 1.77         # the crew multiplied liters by this and treated the result as kg
density = 0.803            # density of the jet fuel (kg/L)
```

**שלב אחר שלב**

הסעיפים כאן זהים לסעיפי העמוד. סעיפים 1 ו-2 הם מה שהצוות **חשב**, וסעיפים 3 עד 5 הם מה שקרה **בפועל**.

**1. כמה ק"ג היו במכלים, לפי הצוות.** הצוות כפל את הליטרים במקדם 1.77 והניח שהתוצאה בקילוגרמים:

```python
crew_kg = fuel_in_tanks_L * factor_used     # 13,597 "kg" according to the crew
```

**2. כמה חסר, לפי הצוות.** ההפרש עד הכמות הנדרשת:

```python
missing_kg = required_kg - crew_kg          # 8,703 "kg" still missing
```

**3. כמה ליטרים נשאבו.** הצוות המיר את החסר חזרה לליטרים, באותו מקדם. ליטרים הם נפח פיזי, ולכן זו הכמות שנשאבה בפועל, גם אם ההמרה שהובילה אליה שגויה:

```python
added_L = missing_kg / factor_used          # 4,917 liters pumped in
```

**4. כמה ליטרים במכלים בהמראה.** הדלק שהיה במכלים ועוד הדלק שנוסף. כאן אין הבדל בין "מה שהצוות חשב" ל"מה שהיה": מדובר בנפח שנמדד ונשאב.

```python
total_L = fuel_in_tanks_L + added_L         # 12,599 liters at takeoff
```

**5. המסה האמיתית.** המסה מתקבלת מהצפיפות של הדלק, לא מהמקדם של הצוות:

```python
fuel_kg = total_L * density                 # about 10,117 kg
```

**מה השתבש**

המקדם 1.77 ממיר ליטרים ל**ליברות** (פאונד), לא לקילוגרמים. ה-22,300 שהצוות חשב שהגיע אליהם היו 22,300 ליברות, כ-45% מהדלק הנדרש. לא הופיעה שום שגיאה: כל המספרים בחישוב היו תקינים, רק היחידה הייתה שגויה. הליטרים (סעיפים 3 ו-4) היו נכונים, והשגיאה נוצרה בהמרה שלהם למסה (סעיפים 1, 2 ו-5).

**תשובות שגויות נפוצות**

- **22,300:** הנחה שהחישוב של הצוות נכון, בלי להמיר בעזרת הצפיפות.
- **12,599:** עצירה בליטרים (סעיף 4), בלי להמיר למסה.
- **6,169:** המרה של הדלק שהיה במכלים לפני התדלוק בלבד (7,682 ליטר), בלי הדלק שנוסף.

**למי שסיים**

`fuel_kg / required_kg` נותן כ-0.45. ליברה אחת היא כ-0.4536 ק"ג, ואכן `1.77 * 0.4536` הוא כ-0.803: הצפיפות של הדלק בק"ג לליטר.
