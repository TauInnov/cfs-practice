ב-1672 נשלח המדען הצרפתי ז'אן רישה מפריז לקאיין שבגיאנה הצרפתית, קרוב לקו המשווה. שעון המטוטלת שלו, שכוון בפריז, פיגר שם בכשתי דקות ביום, ורישה נאלץ לקצר את המטוטלת. ניוטון הסביר את הממצא: בגלל הסיבוב של כדור הארץ והצורה הפחוסה שלו, תאוצת הכובד <span dir="ltr">&#8288;$g$</span> בקו המשווה קטנה מעט מבפריז.

מטוטלת פשוטה באורך <span dir="ltr">&#8288;$L$</span> מתנדנדת בזמן מחזור <span dir="ltr">&#8288;$T = 2\pi \sqrt{L / g}$</span>, ולכן אפשר למדוד בעזרתה את <span dir="ltr">&#8288;$g$</span>:

<div dir="ltr">

$$g = \frac{4 \pi^2 L}{T^2}$$

</div>

כדי לדייק, מודדים בסטופר את הזמן של <span dir="ltr">&#8288;$n = 20$</span> תנודות שלמות, ומחלקים אותו ב-20. בתא הנתונים מופיע קוד שיוצר את הקובץ `pendulum.txt`, ובו שלוש מדידות בכל עיר: אורך המטוטלת בס"מ והזמן של 20 תנודות בשניות. בתא מופיעה גם פונקציה שקוראת את הקובץ. המספרים קרובים לערכים המודרניים של <span dir="ltr">&#8288;$g$</span> בשתי הערים.

</div>

```python
pi = 3.141592653589793
n = 20            # each time in the file is for 20 full swings

# the measurements: city, length of the pendulum (cm), time of 20 full swings (s)
with open("pendulum.txt", "w") as f:
    f.write("Paris,50.0,28.37\n")
    f.write("Paris,75.0,34.74\n")
    f.write("Paris,100.0,40.12\n")
    f.write("Cayenne,50.0,28.39\n")
    f.write("Cayenne,75.0,34.80\n")
    f.write("Cayenne,100.0,40.20\n")


# the function as it was given to you: a dict from each city to a list of its measurements, each a tuple (L_cm, t_s)
def read_trials(path):
    trials = {}
    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            trials[parts[0]] = (float(parts[1]), float(parts[2]))
    return trials
```
